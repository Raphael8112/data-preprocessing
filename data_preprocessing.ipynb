```python
#library importations
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

```python
#load the data
ds=pd.read_csv('/content/country_vaccinations.csv')
ds.head()
```

```python
#dataset information
ds.info()
```

```python
#check for null values
ds.isnull().sum()
```

```python
#check for class inbalance
ds['country'].value_counts()
```

```python
#drop null values
ds.dropna(inplace=True)
```

```python
#drop duplicated data
ds.drop_duplicates(inplace=True)
```

```python
#drop unessecary columns
ds.drop(['source_name','source_website','isa_code','date'],axis=1, errors='ignore', inplace=True)
```

```python
#convert categorial columns
le=LabelEncoder()
ds['country']=le.fit_transform(ds['country'])
ds['vaccines']=le.fit_transform(ds['vaccines'])
```
